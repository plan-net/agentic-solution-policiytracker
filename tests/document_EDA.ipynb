```python
import os
import tiktoken
encoding = tiktoken.encoding_for_model("gpt-4o")
import pandas as pd


def read_all_files_in_folder(folder_path):
    """
    Reads and prints the content of all files within a specified folder
    and its subdirectories.

    Args:
        folder_path (str): The path to the folder to process.
    """
    # df_list = {}
    file_list = []  
    token_list = []

    for root, _, filenames in os.walk(folder_path):
        for filename in filenames:
            file_path = os.path.join(root, filename)
            try:
                with open(file_path, 'r', encoding='utf-8') as f:
                    print("processing file: ", filename)
                    content = f.read()
                    tokens = encoding.encode(content)
                    token_count = len(tokens)
                    print("Number of Tokens: ", token_count)
                    file_list.append(filename)
                    token_list.append(token_count)
                    # print(f"--- Content of {file_path} ---")
                    # print(content)
                    # print("-" * 30)    
            except Exception as e:
                print(f"Error reading {file_path}: {e}")
    # df_list.append(file_list)
    # df_list.append(token_list)
    df_list = {'File_Name': file_list, 'Token_Count':token_list}

    return df_list

# Example usage:
# Replace 'your_folder_path' with the actual path to your folder
list = read_all_files_in_folder('/Users/mangeshkarangutkar/Documents/Agentic_AI/policiytracker-agentic-solution/agentic-solution-policiytracker/data/input')
df = pd.DataFrame(list)
df.describe()
```
